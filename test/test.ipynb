```python
import numpy as np
import rlwe 
import dp
import random


def random_int_list(n: int, low: int, high: int) -> list[int]:
    return [random.randint(low, high) for _ in range(n)]

def modify_k_elements(arr: list[int], k: int, low: int, high: int) -> list[int]:
    n = len(arr)

    if k > n:
        raise ValueError("k cannot be greater than length of array")

    arr = arr.copy()  # 원본 보호

    # k개의 인덱스 선택 (중복 없이)
    indices = random.sample(range(n), k)

    for i in indices:
        new_val = arr[i]
        # 기존 값과 다르게 만들기
        while new_val == arr[i]:
            new_val = random.randint(low, high)
        arr[i] = new_val

    return arr

def print_tables(H_table, V_table):
    print("=== H table (Ph / Mh) ===")
    for i, (Ph, Mh) in enumerate(H_table):
        print(f"Step {i}:")
        print("Ph:", Ph)
        print("Mh:", Mh)
        print()

    print("=== V table (Pv / Mv) ===")
    for i, (Pv, Mv) in enumerate(V_table):
        print(f"Step {i}:")
        print("Pv:", Pv)
        print("Mv:", Mv)
        print()
```
